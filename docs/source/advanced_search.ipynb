# Advanced local index search

NistChemPy no longer ships a prebuilt NIST Chemistry WebBook compound index. Advanced local index examples are being rewritten around the new user-local cache API.

For now, live WebBook search remains available through `nistchempy.run_search()` and `nistchempy.run_structural_search()`. Local index workflows should use `nistchempy.WebBookIndex.from_cache()` once a local index has been generated.
