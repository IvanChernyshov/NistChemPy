# Advanced local index search

NistChemPy no longer ships a prebuilt NIST Chemistry WebBook compound index. Advanced local index workflows now use a user-local cache.

Build or import a local index first, for example:

```bash
nistchempy index build --strategy formula-browser --accept-data-terms
```

or, for a previously extracted local CSV file:

```bash
nistchempy index build --from-csv /path/to/nist_data.csv --accept-data-terms
```

Then load and search it in Python:

```python
import nistchempy as nist

index = nist.get_local_index()
index.search('benzene', fields='name')
```

Live WebBook search remains available through `nistchempy.run_search()` and `nistchempy.run_structural_search()` and does not require a local index.
