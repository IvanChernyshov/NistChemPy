# Local WebBook index

NistChemPy no longer ships a prebuilt WebBook-derived index. Local index search works with a user-generated cache directory or an explicit local CSV file. This example creates a tiny synthetic CSV.


```python
from pathlib import Path
import tempfile
import pandas as pd
import nistchempy as nist

rows = [
    {
        'ID': 'C111111',
        'name': 'Dummybenzene',
        'synonyms': 'Dummy benzol\nDummy cyclohexatriene',
        'formula': 'C6H6',
        'mol_weight': 78.11,
        'inchi': 'InChI=1S/C6H6/dummy',
        'inchi_key': 'DUMMYKEY123',
        'cas_rn': '000-00-0',
        'mol2D': 'https://example.test/mol2d',
        'Mass spectrum (electron ionization)': 'https://example.test/ms',
    },
    {
        'ID': 'C222222',
        'name': 'Dummyethene',
        'synonyms': '',
        'formula': 'C2H4',
        'mol_weight': 28.05,
        'inchi': 'InChI=1S/C2H4/dummy',
        'inchi_key': 'DUMMYKEY456',
        'cas_rn': '111-11-1',
        'mol2D': '',
        'Mass spectrum (electron ionization)': '',
    },
]

tmp = tempfile.TemporaryDirectory()
csv_path = Path(tmp.name) / 'mini_index.csv'
pd.DataFrame(rows).to_csv(csv_path, index=False)
index = nist.get_local_index(csv_path)
index.to_dataframe()
```

The local index object provides simple search, filtering by available sections, and lookup by WebBook ID.


```python
index.search('benzol')[['ID', 'name', 'formula']]
```

```python
index.filter(has_sections=['Mass spectrum (electron ionization)'])[['ID', 'name']]
```

```python
index.get('C111111')[['name', 'inchi_key', 'cas_rn']]
```
